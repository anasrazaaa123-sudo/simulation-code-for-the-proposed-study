```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
indices = {
"λ1(G) = 518 + 230n": lambda n: 518 + 230*n,
"λ2(G) = 646 + 289n": lambda n: 646 + 289*n,
"λ3(G) = 2634 + 1176n": lambda n: 2634 + 1176*n,
"Rλ1(G) = 203 + 93n": lambda n: 203 + 93*n,
"Rλ2(G) = 232 + 105n": lambda n: 232 + 105*n,
"AZI(G) = 920.1667 + 412.083n": lambda n: 920.1667 + 412.083*n,
"ReZG1(G) = 87 + 38.33n": lambda n: 87 + 38.33*n,
"ReZG2(G) = 127 + 59.5n": lambda n: 127 + 59.5*n,
"ReZG3(G) = 3352 + 1510n": lambda n: 3352 + 1510*n
}
rows = []
for n in range(1, 11):
row = {"n": n}
for name, func in indices.items():
row[name] = round(func(n), 4)
rows.append(row)
df = pd.DataFrame(rows)
print("Topological Indices for n = 1 to 10:\n")
print(df.to_string(index=False))
df.to_csv("topological_indices_n1_10.csv", index=False, encoding='utf-8')
df.to_excel("topological_indices_n1_10.xlsx", index=False)
fig, ax = plt.subplots(figsize=(12, 2 + len(df)*0.3))
ax.axis('off')
table = ax.table(cellText=df.values,
colLabels=df.columns,
cellLoc='center',
loc='center')
table.auto_set_font_size(False)
table.set_fontsize(8)
table.scale(1.2, 1.2)
plt.savefig("Topological_Indices_Table.png", dpi=300, bbox_inches='tight')
plt.close()
print("\n Table saved as 'Topological_Indices_Table.png'")
plt.figure(figsize=(10,6), dpi=300)
for col in df.columns[1:]:
plt.plot(df["n"], df[col], marker='o', linewidth=1.8, label=col)
plt.title("Topological Indices Comparison (n = 1–10)", fontsize=14)
plt.xlabel("n", fontsize=12)
plt.ylabel("Index Value", fontsize=12)
plt.legend(fontsize=8)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig("Topological_Indices_Combined.png", dpi=300)
plt.show()
print("\nCombined graph saved as 'Topological_Indices_Combined.png'")
print("Table image saved as 'Topological_Indices_Table.png'")
print(" CSV and Excel files also created.")
```
